Not set up yet? Start with [Environment Setup](https://statfungen.github.io/xqtl-protocol/xqtl_protocol_demo.html) to install the software stack and run a first pipeline on the example data shipped in this repository. Each module below links to its own page, where the inputs, expected outputs and full command interface are documented.

<div class="herorow"><h3 class="hdt">xQTL Analysis Workflow Builder</h3></div>
  <p class="hds">xQTL studies ask how inherited genetic variation changes a molecular phenotype—such as gene expression, splicing, methylation or polyadenylation—and whether that regulatory effect helps explain variation in a complex trait. The map follows that scientific argument from a measured molecular phenotype to a localized regulatory signal and, when appropriate, to evidence connecting the signal with disease biology. Genetic regulation is often tissue- and cell-context-specific, so the relevant path depends on what was measured and where it was measured (<a href="https://doi.org/10.1126/science.aaz1776" target="_blank" rel="noopener">GTEx Consortium, 2020</a>).</p>
</header>
<section class="pipelineintro" aria-labelledby="pipeline-story">
<p class="eyebrow">A practical map of the analysis</p>
<h4 id="pipeline-story">From biological samples to interpretable xQTL discoveries</h4>
<p class="storylead">The pipeline turns a broad biological question—does genetic variation alter a molecular process?—into progressively more specific evidence. Not every project needs every stage: the route should match the molecular phenotype, study design and claim being tested.</p>
<ol class="storygrid">
<li><details><summary><span class="storynum">1</span><b>Define the molecular phenotype</b></summary><p>Turn the assay into biologically interpretable features—expression levels, splice usage, methylation states or polyadenylation usage—in the tissue or cell population relevant to the question. Quality control protects the biological comparison from sample swaps, outliers and poorly measured features.</p></details></li>
<li><details><summary><span class="storynum">2</span><b>Separate inherited regulation from other variation</b></summary><p>Individuals differ because of ancestry, environment, cell composition and technical effects as well as genotype. Accounting for these sources of variation makes the allelic effect easier to interpret as genetic regulation of the molecular phenotype.</p></details></li>
<li><details><summary><span class="storynum">3</span><b>Map and localize regulatory effects</b></summary><p>Association testing identifies loci where genotype tracks the molecular phenotype. Multi-context analysis asks whether an effect is shared across tissues or cell types; fine-mapping narrows correlated variants to credible sets that quantify uncertainty rather than declaring a single causal variant (<a href="https://doi.org/10.1111/rssb.12388" target="_blank" rel="noopener">Wang et al., 2020</a>).</p></details></li>
<li><details><summary><span class="storynum">4</span><b>Relate regulation to complex traits</b></summary><p>Colocalization evaluates whether molecular and trait associations are consistent with a shared causal signal (<a href="https://doi.org/10.1371/journal.pgen.1004383" target="_blank" rel="noopener">Giambartolomei et al., 2014</a>), while TWAS tests whether genetically predicted molecular levels are associated with the trait (<a href="https://doi.org/10.1038/ng.3506" target="_blank" rel="noopener">Gusev et al., 2016</a>). Together with functional enrichment, these analyses prioritize mechanisms but do not by themselves prove mediation or causality.</p></details></li>
</ol>
<p class="storyaction"><b>How to use this page:</b> describe your data and goal, follow the highlighted route, open a module to inspect its real fixture files, and run the displayed <code>sos run</code> command.</p>
</section>

function cmdBlock(d,nb){return `<div class="cmd"><div class="cmdh"><span class="wf">${esc(d.wf)}</span>`
+`<button type="button" class="cpy action" data-cp="${nb}|${d.wf}">Copy</button></div>`
+`<pre>${mark(d.cmd)}</pre></div>`;}
function fmtOf(p){
  for(const f of FMT) if(new RegExp(f[0],'i').test(p))return f;
  return null;
}
function miniTable(cols,rows,note){
  if(!cols||!cols.length)return '';
  return `<div class="prev"><table><thead><tr>`
    +cols.map(c=>`<th>${esc(c)}</th>`).join('')+`</tr></thead>`
    +(rows&&rows.length?`<tbody>`+rows.map(r=>`<tr>`
       +r.map(x=>`<td>${esc(x)}</td>`).join('')+`</tr>`).join('')+`</tbody>`:'')
    +`</table></div>`+(note?`<p class="prov">${note}</p>`:'');
}
function pathTable(paths,nb){
  if(!paths||!paths.length)return '<p class="why">Not given in the commands.</p>';
  return paths.map(p=>{
    const f=fmtOf(p), real=PREV[nb];
    let h=`<p class="pth">${mark(p.split('/').pop())}</p>`;
    if(f){
      h+=`<p class="fmtn">${f[1]}</p>`;
      /* real captured rows win over the format illustration */
      if(real&&/\.bed|phenotype|coord/i.test(p))
        h+=miniTable(real[0],real.slice(1),'First rows, from this notebook&rsquo;s own output.');
      else h+=miniTable(f[2],f[3],
        f[3]&&f[3].length?'Example rows showing the format. Not your data.':'');
    }
    return `<div class="iob">${h}</div>`;
  }).join('');
}
/* I/O follows the selected workflows, not the whole notebook */
function ioFor(nb,wfs){
  const i=[],o=[];
  wfs.forEach(w=>{const v=WIO[nb+'|'+w]; if(!v)return;
    v.i.forEach(x=>{if(!i.includes(x))i.push(x);});
    v.o.forEach(x=>{if(!o.includes(x))o.push(x);});});
  return {i,o};
}

 const rows=p.rows||[],n=p.shownColumns||0,head=rows[0]||[],body=rows.slice(1);
